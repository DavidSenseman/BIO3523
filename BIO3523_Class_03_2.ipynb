# **Hotkeys**

In the last lesson, we looked at some of the most useful slash commands in Hermes Agent. In this lesson we focus on Hermes Agent **hotkeys**. We break down the use of hotkeys into the following 2 categories:

####**1. Navigation & Editing Basics (The "Muscle Memory" Shortcuts)**

These hotkeys will increase your typing speed and efficiency within the terminal interface.

* **Line Navigation:** `Ctrl+A / Ctrl+E` (or `Home/End`) for jumping to line starts/ends.

* **Word Editing:** `Ctrl+W` and `Alt+D` to delete words, `Ctrl+←/→` to jump across words.

* **History & Drafts:** `Esc Esc` to discard drafts and using `↑/↓` to recall history.

####**2. Workflow & Session Control (The Power User Features)**

These hotkeys provide features unique to managing agents and workflows without losing progress.

* **Model Management:** `Ctrl+O` to open the model picker mid-stream without losing the current draft.

* **Live Agents:** `Ctrl+T` to expand active agents and `Ctrl+R / F7` to toggle the preview window.

* **Session Switching:** `Ctrl+X` to open the live session switcher.

* **Shell Interactivity:** `!<cmd>` for running shell commands

###**Word Editing**

* **`Ctrl+W`:** Delete a word (backwards direction).

First, use your mouse to place the cursor on the letter `a` in the word `all` as show in this image:

![__](https://biologicslab.co/BIO3523/images/class_03/class_03_2_image04A.png)

Now use the **`Ctrl+W`** hotkey to delete the the _preceeding_ word `is`.
```text
Ctrl+W
```
You should now see this image:

![__](https://biologicslab.co/BIO3523/images/class_03/class_03_2_image05A.png)

Notice that the word `is` was deleted, not the word `all`. The hotkey `Ctrl+W` deletes the word that **preceeds** the location of the cursor. In other words, the hotkey `Ctrl+W` peforms _backwards_ deletion.

* **`Alt+D`:** Delete a word (forwards direction).

If you want to delete a word in the next word after the current cursor location, you can use the `Alt+D` hotkey.

Make sure your cursor still located on the `a` in the word `all`. Now, delete the word `all` using the `Alt+D` hotkey:
```text
Alt+D
```
Your CLI should now look like this:

![__](https://biologicslab.co/BIO3523/images/class_03/class_03_2_image06A.png)

The hotkey `Alt+D` deletes the word after the cursor location.

* **`Ctrl + ←/→`:** Jump words.

If you want to jump to the next word in a line of text, you can use your arrow keys in combination with the `Ctrl` key. To jump to the next word on the right of the current cursor location, use `Ctrl + →` (right arrow). Conversely, to jump to the left, use `Ctrl + ←` (left arrow).

````python
# @title  Electronic Submission

import urllib.request
import ssl
import time

url = "https://biologicslab.co/BIO3523/backend_code/validate.py?v=" + str(time.time())
ctx = ssl.create_default_context()
ctx.check_hostname = False
ctx.verify_mode = ssl.CERT_NONE
req = urllib.request.Request(
    url,
    headers={
        "Cache-Control": "no-cache, no-store, must-revalidate",
        "Pragma": "no-cache",
        "Expires": "0"
    }
)
with urllib.request.urlopen(req, context=ctx) as r:
    exec(r.read().decode("utf-8"))
main()
````
