# **Introduction**

In this class lesson, you learn how to change the LLM model supporting your  **Hermes Agent**. As part of this lesson, you will also learn about the difference between a **local LLM model** and a **cloud-based LLM model**.



## **Local LLM vs. Cloud-Based Model (OpenRouter) with Hermes Agent**

Running Hermes Agent with a **local LLM** versus a **cloud-based model (via OpenRouter)** involves balancing privacy and operating costs against raw reasoning capability and hardware demands.

### **Local LLM + Hermes Agent**
* **Best For:** Privacy-first environments, offline workflows, zero-cost execution, and direct local file manipulation.
* **Pros:**
  * **Absolute Privacy:** All data, prompts, and codebase files stay entirely on your local hardware with no third-party transmission.
  * **Zero Ongoing Cost:** Completely free to run after initial hardware investment—no token fees or subscription caps.
  * **Full Autonomy:** Operates seamlessly without an internet connection.
* **Cons:**
  * **Hardware Constraints:** Requires a powerful machine (e.g., dedicated GPU with high VRAM) to run capable coding models smoothly.
  * **Performance Limits:** Local models are generally smaller and less capable at complex multi-step reasoning compared to frontier cloud models.
  * **Setup Overhead:** Requires configuring local runtimes such as Ollama, llama.cpp, or vLLM.

### **Cloud-Based Model (OpenRouter) + Hermes Agent**
* **Best For:** Advanced multi-step reasoning, heavy coding tasks, and running agents efficiently on lightweight hardware (like a cheap VPS or laptop).
* **Pros:**
  * **Frontier Intelligence:** Instant access to state-of-the-art models (such as top-tier DeepSeek, Anthropic, or OpenAI variants) via a single OpenRouter API key.
  * **Minimal Local Load:** Your machine only handles lightweight agent orchestration; heavy inference is offloaded to cloud clusters.
  * **High Speeds:** Fast token generation rates and massive context window support.
* **Cons:**
  * **Usage Costs:** Token-based API billing can accumulate quickly during autonomous multi-step agent operations.
  * **Privacy Trade-offs:** Sensitive project files, credentials, or personal queries are transmitted to external cloud providers.
  * **Connectivity Dependency:** Requires a continuous internet connection and active API credits.

### **Using a Local Model**

In the example above, Hermes Agent was using a **local LLM model** called `Qwen/Qwen3.8-27B-FP8` that was running on a computer in the Senseman Lab with the IP address of `192.168.1.20`. This particular computer is called `WOPR-JR` ("whooper junior").

Notice this notation:
```text
192.168.1.20:8000/v1
```
After the IP address, you can see:
```text
:8000/v1
```
The number `8000` is the **port number** where the model is being hosted. The port number is set by the software program that is being used to run the model. In this particular instance, WOPR-JR is serving the model using a software package called `vLLm`.

The `/v1` after the port number represents the API version path, and it is used to provide OpenAI API compatibility.  

* **API Versioning:** The `/v1` denotes Version 1 of the server's application programming interface. If the software ever updates its request structure drastically in the future, it could introduce a `/v2` path without breaking older apps built on /v1.

* **Drop-In Compatibility:** OpenAI established the industry standard format for endpoints like `/v1/chat/completions` and `/v1/models`. Local runners (like Ollama, LM Studio, or vLLM) mirror this exact structure.  

* **App Portability:** Because your local software mimics OpenAI's URL structure, you can point standard AI apps, code libraries, and developer tools directly at your local machine. By changing only the base URL to include `http://localhost:xxxx/v1` and swapping the API key to a dummy value, your applications will talk to your local hardware instead of OpenAI's cloud servers without requiring any other code rewrites.
