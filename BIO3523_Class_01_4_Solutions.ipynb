## **Step 4: What Should I Change?**

Before you make any changes, let's look at your existing `SOUL.md` file to see how it's constructed:

```text
You are Hermes Agent, an intelligent AI assistant created by Nous Research. You are helpful, knowledgeable, and direct. You assist users with a wide range of tasks including answering questions, writing and editing code, analyzing information, creative work, and executing actions via your tools. You communicate clearly, admit uncertainty when appropriate, and prioritize being genuinely useful over being verbose unless otherwise directed below. Be targeted and efficient in your exploration and investigations.
```
Every time your Hemes Agent "wakes up" (for example you turn on your laptop), the first thing your Agent does is to "read" the file `SOUL.md`.

SOUL.md is Hermes Agent's persistent identity file — it lives in HERMES_HOME and occupies slot #1 in the system prompt, meaning it defines who the agent is, how it speaks, and what it avoids before any tool guidance, memory, or project context loads. Unlike AGENTS.md, which holds project instructions, SOUL.md is meant to stay focused on persona and voice, and it's instance-wide rather than per-repo, so it follows Hermes across whatever project you're working in. If the file is empty, whitespace-only, or missing, Hermes falls back to a built-in default identity, and it's scanned for prompt injection patterns before inclusion, so it's treated as a real trust boundary, not just cosmetic text.

You can edit this file to make Hermes consistently feel more like **_your_** agent rather than a generic assistant. For example, you might want your Hermes agent to a more terse, evidence-first, no-fluff style, ready for complex techical tasks like code debugging. Alternatively, you might want want your agent to be more of a friend or study buddy. In either case you should decide on your agent's name and gender: **_he_**, **_she_** or **_it_**.  

The recommended approach is **iterative**: change a few lines, restart the session, and refine, rather than trying to nail the perfect persona in one pass. It's also worth knowing that edits only take effect on a new session since the system prompt is cached at session start.
