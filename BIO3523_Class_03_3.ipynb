# **Skills (Bundled)**

### **Overview of Skills in Hermes Agent**

In Hermes Agent **skills** are reusable, modular operating instruction packages (`SKILL.md` files accompanied by optional scripts, references, and assets) that teach the agent how to perform specific, multi-step workflows. Instead of overloading the system prompt or forcing you to re-explain a complex process every time, Hermes dynamically loads these procedural memory packages only when needed—saving tokens while ensuring consistent, high-quality execution across tasks spanning coding, research, DevOps, and task automation.

---

### **Most Important Bundled Skills**

* **`skill-manager`**
Handles the creation, patching, updating, and maintenance of agent skills, allowing Hermes to write its own procedural memory when a successful multi-step workflow is discovered.
* **`web-research`**
Orchestrates structured web crawling, source verification, and content aggregation, instructing the agent on how to filter out low-quality pages and synthesize raw web data into clean summaries.
* **`git-workflow`**
Provides standardized rules and checklist patterns for managing version control, staging changes, writing clean commit messages, handling branches, and resolving merge conflicts safely.
* **`mcp-connector`**
Facilitates secure connection to Model Context Protocol (MCP) servers, letting the agent discover, filter, and execute external tools safely.
* **`cron-scheduler`**
Automates recurring tasks and scheduled jobs, allowing the agent to run periodic briefs, monitoring tasks, or maintenance checks and deliver results directly to messaging channels.
* **`subagent-spawn`**
Instructs Hermes on how to delegate workstreams by spinning up isolated, parallel subagents to tackle heavy, multi-pronged tasks concurrently.
* **`security-audit`**
Enforces strict safety boundaries and command approval policies, ensuring the agent pauses or rejects dangerous shell commands, unvetted package installations, or unauthorized file modifications.
* **`document-formatter`**
Guides the formatting, conversion, and structuring of output files (such as markdown reports, CSVs, or code files) to match exact styling guidelines before saving or exporting.
* **`error-troubleshooter`**
A systematic debugging workflow that parses stack traces, runs diagnostic commands (`hermes doctor`), and analyzes error logs to safely troubleshoot failed scripts or broken execution paths.
