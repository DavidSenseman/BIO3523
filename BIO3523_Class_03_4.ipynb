# **Skills (Downloaded)**

### **Overview of Skills in Hermes Agent**

In the previous lesson we looked at some of the skills that come bundled with the Hermes Agent installation. But there are a large and growing number of skills that you can download from the internet. Here is a comparison between bundled and downloaded skills.

---

### **Bundled Skills**

**Advantages**
*   **Out-of-the-box readiness:** Bundled skills are copied directly from the Nous Research repository into your `~/.hermes/skills/` directory on a fresh install, allowing you to execute common workflows immediately.
*   **Vetted reliability:** Because they are officially maintained alongside the core agent, they are tested to work seamlessly with Hermes' 60+ built-in tools and platform configurations.
*   **Automatic synchronization:** The system keeps bundled skills seeded and synced during runtime and through `hermes update` commands, ensuring they stay current unless you explicitly run the `hermes skills opt-out` command.

**Disadvantages**
*   **Limited scope:** They are confined to broadly applicable, officially supported tasks—such as MLOps, GitHub operations, diagramming, and note-taking—meaning highly specialized workflows will not be available natively.
*   **Update dependency:** Improvements and new additions are tied to the official Nous Research release cycle rather than rapid, decentralized community iteration.

### **Downloaded Skills (Skills Hub & Community)**

**Advantages**
*   **Extensive variety:** Downloading from external hubs provides access to a massive, community-contributed library of workflows that go far beyond the base installation.
*   **Open standard compatibility:** Community skills adhere to the `agentskills.io` open standard and the `SKILL.md` format, making them highly portable and easy to drop into your environment with a single command.
*   **Niche specialization:** You can leverage the specific problem-solving workflows of other developers who have already spent time refining agent behavior for obscure or highly technical edge cases.

**Disadvantages**
*   **Security risks:** Skills dictate how the agent uses its tools, manages files, and executes commands within its environment. Running unvetted community scripts inherently carries higher security and privacy risks.
*   **Variable quality control:** Third-party skills may be poorly optimized, consume excessive context tokens, or break when Hermes Agent updates its Tool Gateway or underlying architecture.

### **Most Important Downloadable Skills**

The open `agentskills.io` standard (which Hermes Agent supports natively) provides access to thousands of community and scientific skills. Because the One-Health Initiative requires integrating human, veterinary, and environmental health data, the most useful skills focus on bioinformatics, epidemiology, advanced research workflows, and system visualization.

Here are the 10 most useful skills a One-Health student can download for their Hermes Agent:

#### **Specialized Scientific Skills**
1. **`ddkg.skill` (Data Distillery Knowledge Graph):** Essential for bioinformatics. This compositional skill guides your agent in translating complex biomedical and bioinformatics questions into Cypher queries. It integrates over 180 ontologies and 40 genomics datasets, allowing you to cross-reference human phenotypes, clinical variables, and cross-species genomic overlap.
2. **Veterinary Epidemiologist (Grade A):** A highly specialized workflow skill that provides your agent with the exact methodologies needed for outbreak investigation. It guides the agent through establishing case definitions, confirming lab methods (like PCR and sequencing), formulating hypotheses (wildlife, fomites, aerosol), and designing analytic cohort studies.
3. **K-Dense Genomics Module:** Part of K-Dense's massive library of 163 scientific research skills, this module equips the agent with domain-specific knowledge and workflows for processing genomic datasets and understanding genetic sequencing workflows.
4. **K-Dense Clinical & Drug Discovery Modules:** Also from the K-Dense library, these skills provide specialized frameworks for analyzing clinical research data and pharmacological interactions, bridging the gap between animal models and human therapeutics.

#### **Deep Research & Literature**
5. **Deep Research (Universal Team):** Instead of a single agent, this skill deploys a 13-agent pipeline designed for rigorous academic research. It features 8 distinct modes, including systematic reviews with optional meta-analysis, three-way literature scans, and fact-checking—perfect for synthesizing environmental, animal, and human health literature.
6. **Headless Browser Automation (Playwright):** Standard LLMs can't browse the live web effectively. This skill gives your Hermes agent sandboxed WASM QuickJS and Playwright capabilities, allowing it to autonomously navigate, read, and pull live epidemiological data from global health registries or environmental monitoring sites.

#### **Data Visualization & Engineering**
7. **Beautiful Architecture Diagrams:** One-Health often involves modeling complex transmission pathways, ecosystems, or data-flow pipelines. This skill allows the agent to take your raw system descriptions and generate verifiable, interactive architecture, workflow, and lifecycle diagrams rendered locally as HTML/SVG.
8. **Hugging Face Ecosystem Plugin:** Maintained by Hugging Face, this skill plugs your agent directly into their ecosystem. It allows the agent to autonomously search for machine learning models, access public datasets, and spin up inference endpoints—vital for applying ML to disease tracking and predictive modeling.
9. **Agent-Skills-for-Context-Engineering:** Complex bioinformatics queries often degrade the agent's memory. This foundational skill by Muratcan Koylan manages context compression and memory systems, ensuring your agent doesn't lose the thread when analyzing massive, multi-species datasets over long sessions.
10. **Hybrid Code Review (Open Code Review):** When writing Python or R scripts for statistical analysis, reproducibility is critical. This skill provides deterministic pipelines and LLM-driven line-level comments to audit your code, ensuring your data-processing scripts are thread-safe, efficient, and free of syntax vulnerabilities before you run them against sensitive health data.

####**The Scenario: "The San Antonio Feed Store Cluster"**

**Background:**

Over the last 14 days, several backyard chicken owners in suburban San Antonio have reported a sudden onset of severe respiratory signs in their flocks, including nasal discharge, facial swelling, and a sharp drop in egg production. Two owners reported purchasing pullets from the same local feed store swap just before the symptoms began.

**How to Use the Skill:**

Initiate the Hermes Agent with the Veterinary Epidemiologist skill loaded, and feed it the following starting prompt:

```text
I am initiating a field outbreak investigation for an unexplained respiratory disease cluster affecting backyard poultry flocks in San Antonio, Texas.

Here is our initial field report:
- 5 separate premises have reported symptomatic birds in the last 14 days.
- Clinical signs: facial edema, purulent nasal discharge, sneezing, and a 40-60% drop in egg production.
- Mortality is low, but morbidity is nearly 80% within affected coops.
- 2 of the affected premises recently introduced new pullets acquired from a local poultry swap event.
- Preliminary suspicion: Infectious Coryza (Avibacterium paragallinarum) or Mycoplasma gallisepticum (MG).

Please walk me step-by-step through the standard epidemiological field protocol for an outbreak investigation. Start by establishing our lab confirmation strategy and defining what constitutes a "case."
````

Because the Veterinary Epidemiologist skill forces the agent into a highly structured, 8-step field protocol, it will stop the LLM from simply guessing the disease and instead force the user to do the epidemiological work.

The agent will guide the student through the following progression:

**Verify Diagnosis:** The agent will first instruct the student not to act on the "preliminary suspicion" alone. It will ask the student to define the exact lab method needed for confirmation (e.g., asking the student to "order" PCR testing from choanal cleft swabs rather than relying on clinical signs).Case Definition & Line List: It will prompt the student to build a line list. The student will have to define the parameters (e.g., "Any chicken in Bexar County exhibiting facial swelling and nasal discharge between May 1 and May 15"). The agent will request data like flock size, vaccination status, and movement history.   

**Epidemic Curve:** The agent will ask for the exact onset dates for the 5 premises to map the epidemic curve and attempt to identify the true index case.   

**Hypothesis Formulation:** The agent will prompt the student to look beyond the "new pullets" and formulate alternative hypotheses (e.g., wild bird interaction due to open-air coops, shared water sources, or fomite transmission on the shoes of owners visiting the feed store).   

**Study Design & Analysis:** This is where the computational biology aspect shines. The agent will instruct the student to design an analytic study—likely a case-control study comparing the 5 affected premises against 15 unaffected backyard flocks in the same area, using Odds Ratios (OR) to calculate the statistical link to the feed store swap.

How the Hermes Pipeline Would Handle This:

**Task Delegation:** The primary agent parses the prompt and assigns specific search parameters to sub-agents dedicated to the three distinct literature pools (environmental, animal, and human health).

**Data Extraction & Three-Way Scan:** The specialized agents independently retrieve and summarize abstracts, datasets, and methodologies from their assigned domains.

**Fact-Checking:** Before compiling the research, the designated fact-checking agents cross-reference the extracted transmission statistics to flag any discrepancies, contradictory studies, or hallucinated data points.

**Synthesis:** A final agent merges the verified data into a cohesive, rigorously cited literature review suitable for a BIO 3523 project, highlighting the intersections between the three health domains.
