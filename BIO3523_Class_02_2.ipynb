## **Hermes Agent with a Local Model**

The "brain" of your Hermes Agent is a Large Language Model (LLM), or more simply a **model**. This model can be either `cloud-based` or `local`.

### **Cloud-based Models**

Your Hermes Agent can access a cloud-based model such as `deepseek/deepseek-v4-pro-0813` through OpenRouter using your OpenRouter API key. Your request is typically handled dynamically by one of OpenRouter's integrated third-party host providers or infrastructure partners globally. Because DeepSeek models are open-weights, your tokens may be processed on hardware managed by cloud providers and hosting partners (such as Baseten, Fireworks, Azure, or other global inference providers) rather than exclusively by DeepSeek's own data centers.

#### **Hardware**
Full-scale frontier models like DeepSeek-V4-Pro require massive server nodes because their raw weight sizes reach hundreds of gigabytes to over a terabyte (even when packed or pre-quantized).   

* **GPUs:** Multi-node clusters or single high-density enterprise nodes packed with NVIDIA flagship accelerators, such as NVIDIA H100, H200, or Blackwell (e.g., GB200/GB300 systems).

* **VRAM/Memory:** 900 GB to over 1 TB of aggregate high-bandwidth memory (HBM) just to hold the model weights, plus substantial extra headroom for massive context windows (such as 1M-token contexts).

* **Infrastructure:** Server-grade multi-core CPUs (AMD EPYC or Intel Xeon), hundreds of gigabytes to terabytes of system RAM, ultra-fast NVMe storage arrays in RAID configurations, and data center-grade InfiniBand or high-speed Ethernet (100–200 Gbps+) for fast tensor/pipeline parallelism across multiple GPUs.

### **Local Models**

A **local model** is an LLM model whose weights and inference engine reside and execute entirely on your own local hardware (such as a dedicated workstation, local server, or desktop computer) rather than being hosted on remote cloud infrastructure or accessed via external APIs.

Students in BIO 3523 can access two local models that are running on two computers within in the same local area network (LAN) as the computer that is hosting the LXC Linux Container. These computers are called `WOPR` (IP 192.168.1.83) and `WOPR-JR`(IP 192.168.1.20). `WOPR` is a Dell ProMax with GB10 (the Dell version of the Nvidia DGX Spark) while `WOPR-JR` is AMD Threadripper Pro computer with two Intel ARC Pro B70 video cards, each with 32 GB of VRAM.
    
#### **Advantages of Local Models**

**Low Operational Costs:** Using a local model is obviously attractive since you don't have to pay for tokens. If you don't count the extra electricity consumed by the computer running the model, the local model is basically **free** to run!

**Data Privacy & Sovereignty:** All prompts, context windows, and generated tokens remain strictly on-device, ensuring complete security and zero third-party data logging.

When you use a cloud-based model, you are sending your data to the company hosting the model. These companies use your data, primarily but not exclusively to train their next generation of LLMs. This is true of both companies located in the US (e.g. Anthropic, OpenAI, Google) as well as foreign companies that are mostly located in China (e.g. Qwen, and Deepseek). How worried you should be about sending your data to a private company will obviously depend upon the nature of the data as well as your personal level of privacy concerns. Importantly, there are several important types of **protected data** that should _never_ be analyzed using a cloud-based model including student records (FERPA protected) and health recorld (HIPPA protected).

#### **Disadvantages of Local Models**

**Equipment Costs:** The largest barrier to using a local model is the cost of the computer hardware. While you can run a local model on a "gaming PC" with a decent NVIDIA graphics card or on a modern Apple laptop, Mac mini or Mac Studio, your experience may--or may not--be positive depending upon how much money you spent. For example, a gaming PC with a "top-of-the-line" Nvidia RTX 5090 GPU can run some "good" models at a relatively fast pace (> 30 tokens/sec) while a low-end, older Nvidia RTX 3050 will struggle to run any model. The new Mac Studio with M5 Ultra and 512 GB of memory can run some very large models reasonably fast but an older laptop won't really run any models. Expect to pay at least \$3000 for a computer and graphics card that can run a local model.

**Software Configuration:** Even if you have adequate hardware, configuring the software to run a local model (i.e. a local interence engine) can be challenging. The easiest software that you can use is called `ollama` while the most difficult is called `vLLM`. In between these two extremes is a software package called `llama.cpp`. As you might imagine, more complex software like `llama.cpp` and `vLLM` will give you more control over how the model runs and will probably run it faster than it would just using `ollama`. The good news is that all these software packages are open-sourced and free to run.



## **Part A: Comparing Cloud-based and Local Qwen Models**

**Qwen** (short for Tongyi Qianwen) is a comprehensive family of large language models and multimodal AI systems developed by Alibaba Cloud. Ranging from ultra-compact lightweight models to massive enterprise-grade mixture-of-experts (MoE) architectures, the series covers text generation, coding, mathematics, vision-language tasks, and agentic tool use.   

#### **Why Qwen Models Are So Popular**

* **Exceptional Open-Weight Performance:** Qwen models consistently match or exceed the performance of many proprietary frontier systems and rival open-source alternatives across standard evaluation benchmarks for reasoning, coding, and multilingual understanding.

* **Granular Scaling & Efficiency:** Available in dozens of precise parameter configurations (from sub-billion scales to hundreds of billions) and offering both dense and memory-efficient Mixture-of-Experts (MoE) designs, developers can easily match a model to their exact hardware constraints.   

* **Massive Global Ecosystem Adoption:** Qwen has grown into one of the most widely downloaded and utilized open-source model families globally, forming the bedrock for countless fine-tuned derivative models, research projects, and production pipelines.   

* **Advanced Multimodal & Long-Context Support:** The family excels in specialized domains, featuring natively integrated vision-language capabilities (Qwen-VL), robust coding/math variants, and massive context handling well-suited for complex agent workflows.

## **Vibe Coding a Token Measuring App**

Now that you have switched to `Qwen3.8-flash`, let's use it for some vibe coding.

Send this prompt to your Hermes Agent:

```text
Build a command-line app called `llm-bench` that measures token usage and
generation throughput for a language model and keeps a history so different
models can be compared.

MODEL SELECTION (ask at startup):
  A) Your own current model — the model you, Hermes Agent, are running as.
  B) A local model — prompt the user for a base URL and port
     (e.g. http://192.168.1.50:11434) and, if needed, a model name.
Handle both the same way in code: an OpenAI-compatible chat-completions
endpoint — POST {base}/v1/chat/completions with a JSON body containing
`model`, `messages`, `temperature`, `max_tokens`, and expect a `usage` object
in the response. For option A, fill in your own endpoint/model name as a preset
so one code path serves both. Build a model key = model name + endpoint host;
use it to identify runs in the history.

TEST PROMPTS (fixed — identical text every run):
  1. code generation:      "Write a Python function using the Sieve of
     Eratosthenes that returns the first N primes. Include a docstring and
     a couple of doctests."
  2. scientific knowledge: "Explain how the sodium–potassium pump maintains
     the resting membrane potential of a neuron."
  3. general knowledge:    "Summarize the main causes of the fall of the
     Western Roman Empire."

GENERATION SETTINGS (identical for every call, for comparability):
  temperature = 0, max_tokens = 512.

MEASUREMENT PROCEDURE:
  1. WARMUP: send each of the 3 prompts once and DISCARD results and timings
     (this loads the model and warms caches).
  2. Then run 3 MEASURED runs. In each run, for each of the 3 categories:
       - record wall-clock start, send request, record end;
       - from the response `usage` field read prompt_tokens (tokens IN) and
         completion_tokens (tokens OUT);
       - throughput tk/s = completion_tokens / elapsed_seconds.
  3. For each category, average tokens_in, tokens_out, and tk/s across the 3
     runs. Also compute an overall average tk/s across all 9 measured calls.
  If the endpoint returns no `usage`, fall back to counting with tiktoken (or
  estimate) and note that in the output.

STORAGE:
  After every use, append one record to `benchmark_history.json` (create if
  absent). Each record: timestamp, model key (name + endpoint), settings
  (temperature, max_tokens), per-category {avg tokens_in, avg tokens_out,
  avg tk/s}, and overall avg tk/s.

COMPARISON:
  At the end of a run, load the history. If the current model key differs from
  the most recently stored run's key, print a comparison table: one row per
  distinct model, columns for overall avg tk/s and per-category avg tk/s (plus
  tokens in/out), with the current run highlighted. Always print the current
  run's own results table first.

OUTPUT: clean terminal tables (use `rich` if available). Handle unreachable
endpoints, timeouts, and HTTP errors with clear messages. One file if practical.

```

### **Turn Your App into a Skill**

A Hermes Agent **skill** is a modular, on-demand capability package (centered around a SKILL.md file) that teaches the agent how to execute a specific, multi-step workflow—such as managing GitHub pull requests, running debugging sequences, or interacting with specific APIs.   

#### **Why They Are Useful**

* **Context Efficiency:** Instead of permanently bloating the model's context window with heavy instructions for every possible task, skills use a progressive disclosure model (loading lightweight indexes first and fetching full instructions only when a matching task arises).

* **Procedural Consistency:** They lock down proven operating procedures, tool sequences, and verification steps so the agent executes specialized tasks reliably without needing exhaustive step-by-step prompting every time.   

* **Modular Extensibility:** Users can seamlessly plug in bundled or custom skill packages across categories like DevOps, research, and software development, allowing the agent to expand its functional repertoire cleanly over time.

### **Discussion**

**There is no universal winner.** The cloud model was about twice as fast in Part A, but the local model was about twice as fast in Part B. "Local versus cloud" turns out to be the wrong question. Speed depends on the specific model, how it is compressed, and the hardware it runs on.

**Why was the local Nemotron so fast?** Generating each token requires the hardware to read the model's *active* weights from memory, so fewer and smaller active weights mean faster generation. The names contain clues:

* `Qwen3.8-27B` is a **dense** model, so all 27 billion parameters are used for every token.
* `Nemotron-3.5-Lightning-30B-A3B` is a **Mixture-of-Experts (MoE)** model. It has 30 billion parameters in total, but only about 3 billion are active per token (the "A3B"). The cloud Nemotron is `120b-a12b`, with about 12 billion active.
* The local models are also compressed differently. `FP8` stores each weight in 8 bits, while `NVFP4` uses only 4 bits, so less data has to be read per token.

**Why was the local Qwen so consistent?** Its speed was essentially identical across all three prompt types (26.92 to 26.98 tk/s). A local server dedicated to one user does the same work at the same speed every time. The cloud models varied more (for example, cloud Nemotron ranged from 45.87 to 69.89 tk/s). A cloud request travels over the internet, may wait in a queue, and shares hardware with other users. Our timer runs from request to response, so this overhead is included in the measurement.

**What this benchmark does *not* tell you.** Please keep these limits in mind:

* **Speed is not quality.** A faster model is not a better model. We measured tokens per second and said nothing about the correctness of the answers.
* **The pairs are not identical models.** They differ in size, architecture, and compression, so each comparison mixes several variables at once.
* **Both Nemotron models hit the 512-token limit** (tok out = 512.00), so their answers were cut off. The reported token counts may also include hidden "reasoning" tokens, so tk/s is not the same as words read per second.
* **Input tokens are a subtle clue.** Nemotron used identical input token counts in the cloud and locally (37.67), but Qwen did not (83 vs. 74). This suggests the two Qwen providers wrapped the same prompt in slightly different templates.

**What about cost?** When this lesson was prepared, the total OpenRouter charges were **less than \$0.35 for the Qwen model** and **less than \$0.06 for the Nemotron model**. These totals cover everything the agent did, including writing the app and the skill, not only the benchmark runs. The local models charge nothing per token, but the hardware behind them costs at least \$3000 (plus electricity). At roughly \$0.41 for the whole lesson, you would have to complete thousands of lessons like this one before buying hardware paid for itself. For occasional use, the cloud is much cheaper. Local models make sense when you have sustained heavy usage or need offline access. They also make sense when **privacy** matters, such as with FERPA-protected student records or HIPAA-protected health data, which should never be sent to a cloud model.
