# **Introduction**

In this class lesson, you will attach your **Hermes Agent** to the social networking application **Discord**. Discord is a real-time communication platform that brings people together through text channels, voice chat, video, and direct messaging. Originally built for gamers, it has evolved into a primary hub for developer communities, AI research groups, open-source projects, and remote teams.

Information on Discord is organized into **Servers** (communities), which contain **Channels** dedicated to specific topics, as well as **Direct Messages (DMs)** for one-on-one conversations.


### **Why Connect Your Hermes Agent to Discord?**

**Hermes Agent** (developed by Nous Research) is an autonomous, self-improving AI agent capable of managing memory, creating reusable skills, executing tools, and running long background tasks. Integrating Hermes with Discord extends its capabilities in several key ways:

#### **Cross-Platform Accessibility**
* **Access Anywhere:** Chat with your agent on mobile, desktop, or web without needing to open a terminal or run a local desktop GUI.
* **DM or Channel Access:** Interact via private DMs for personal assistance, or mention the agent in server channels for community tasks.

#### **Team Collaboration & Community Assistance**
* **Shared Knowledge Base:** Hermes can act as a technical assistant, answering team queries, summarizing discussions, or searching documentation directly inside your server.
* **Role & User Access Control:** Fine-tune permissions (such as `DISCORD_ALLOWED_USERS` or `DISCORD_ALLOWED_ROLES`) so only authorized group members can trigger specific agent actions or tools.

#### **Voice Channels & Audio Capabilities**
* **Live Voice Interaction:** Hermes supports joining Discord voice channels for hands-free conversations using continuous Speech-to-Text and Text-to-Speech (STT/TTS).
* **Spoken Audio Replies:** Have Hermes send voice note responses alongside standard text messages.

#### **Background Automation & Alerts**
* **Scheduled Cron Workflows:** Set up background tasks or scheduled jobs that send automated updates, alerts, or summaries directly into specified Discord channels.
* **Asynchronous Subagents:** Spawn subagents for heavy tasks on a remote VPS or cloud instance while receiving real-time status updates in Discord.

## **Step 4: Enable Privileged Gateway Intents** ⚠️ *Critical step*

Without the correct intents enabled, your bot will connect to Discord but **cannot read what users type**. This is the #1 reason "online but unresponsive" bots exist.

Scroll down to **Privileged Gateway Intents** and toggle **ON**:

- **Server Members Intent** — Required. Resolves usernames so Hermes can identify *who* is messaging it and enforce the allowed-users list.
- **Message Content Intent** — Required. Lets Hermes read the actual text of messages.

![__](https://biologicslab.co/BIO3523/images/class_01/class_01_3_image08B.png)

> **Note:** Without *Message Content Intent*, the bot joins the server but every message text arrives empty — it literally cannot see what you typed.

Click **Save Changes** at the bottom.

> **Verification tip:** If your bot is ever online but never responds, this intent is almost certainly off. Re-check it first.

## 🛠️ Troubleshooting Quick Reference

| Symptom | Likely cause / fix |
|---|---|
| Bot online but won't respond | **Message Content Intent** is off — re-enable it (Step 4). |
| `"Disallowed Intents"` error on startup | Intents not enabled in the Developer Portal. |
| Bot can't see a specific channel | Missing `View Channels` permission, or the channel is in `DISCORD_IGNORED_CHANNELS`. |
| `403 Forbidden` errors | Bot lacks required permissions (re-check Step 6). |

````python
# @title Exercise 1: Verify Your Discord Bot Token

import re
import gc
import base64
from getpass import getpass
from IPython.display import clear_output, Image, display

# ── INPUT SECTION ──────────────────────────────────────────────────
bot_token_ex = getpass("Use CTRL-V or CMD-V to paste your Discord bot token here and hit Enter: ")
# ─────────────────────────────────────────────────────────────────

# Scan token

# A Discord bot token is 3 dot-separated URL-safe Base64 segments,
# and the first segment decodes to the bot's numeric user ID.
pattern_ex = re.compile(r'^[A-Za-z0-9_-]+\.[A-Za-z0-9_-]+\.[A-Za-z0-9_-]+$')
stripped_token_ex = bot_token_ex.strip()
is_valid_ex = bool(pattern_ex.match(stripped_token_ex))

if is_valid_ex:
    first_seg = stripped_token_ex.split('.')[0]
    padded = first_seg + '=' * (-len(first_seg) % 4)
    try:
        user_id = base64.b64decode(padded, validate=True).decode()
        is_valid_ex = user_id.isdigit()
    except Exception:
        is_valid_ex = False

if is_valid_ex:
  print("Your Discord bot token is Valid")
  display(Image(url="https://biologicslab.co/BIO3523/images/class_01/congratulations-congrats.gif"))
else:
  print("Your Discord bot token is not Valid. \nPlease try again.\n")
  print("Hints:")
  print("Copy-and-paste your Discord bot token from your text file into the entry box and hit Enter.")
  print("If you are still having problems, please ask your Instructor for help.")
````
