# **Introduction**

In this class lesson, you will attach your **Hermes Agent** to the messaging application **Telegram**. Telegram is a fast, cloud-based messaging platform available on your phone, tablet, laptop, and the web. Alongside ordinary chats, Telegram has a first-class **Bot API** — a built-in way for programs (like your Hermes Agent) to send and receive messages through an automated account called a **bot**.

Unlike some platforms, you do not need to write any networking code to use it: **Hermes Agent speaks Telegram natively**. You simply create a bot, hand its token to Hermes, and start the gateway.

### **Why Connect Your Hermes Agent to Telegram?**

**Hermes Agent** (developed by Nous Research) is an autonomous, self-improving AI agent capable of managing memory, creating reusable skills, executing tools, and running long background tasks. Connecting Hermes to Telegram extends its reach in several ways:

#### **Cross-Platform Accessibility**
* **Access Anywhere:** Chat with your agent from your phone, tablet, or laptop without opening a terminal.
* **Private by Default:** You talk to *your own* bot in a private chat — no shared server to join.

#### **Voice & Media**
* **Voice Memos:** Send a voice message and Hermes will transcribe it automatically, so you can talk instead of type.
* **Images & Files:** Send pictures and documents directly to your agent.

#### **Background Automation & Alerts**
* **Scheduled Cron Workflows:** Hermes can run background jobs on a schedule and deliver the results straight to your Telegram chat — you will use this in Module 2.

## **The 2 Reasons You Will Like Hermes/Telegram**

Here are the two reasons that you will especially like having your Hermes Agent connected to Telegram:
1. **Talk instead of Type**: If you like talking more than typing, just send your Hermes Agent a Telegram voice memo. Your agent will understand your voice — even in a non-English language — and can reply in the language of your choice.
2. **Use Your Phone, not your Laptop:** Because Telegram runs on your phone as well as your laptop, you can reach your Hermes Agent anytime, anywhere — as long as your Hermes gateway is running.

## 🛠️ Troubleshooting Quick Reference

| Symptom | Likely cause / fix |
|---|---|
| `@BotFather` won't accept your username | Usernames must be unique and **end in `bot`** (e.g. `myname_bot`). Try another. |
| Bot never replies | The gateway isn't running — start it with `hermes gateway` (Step 5). |
| `Unauthorized` error on gateway start | The token is wrong or was revoked. Get a fresh one from @BotFather with `/token` and re-run `hermes gateway setup`. |
| Bot ignores you specifically | You set an allowed-users list that doesn't include your username. Re-run setup and leave it empty, or add yourself. |
| Cron results never arrive | You haven't set a home channel — send `/sethome` to your bot (Step 6). |

# **Exercise 1 : Verify Your Bot's Token**

This assignment is complete once you have obtained a valid **Telegram bot token**. The cell below simply verifies that your token has the correct form.

When you run the code, a textbox will appear. Copy your bot token from your text file, click inside the box, and paste it with **`CTRL-V`** (Windows) or **`CMD-V`** (Mac). To keep it secure, you will **not** see the text as you type it. While your cursor is still inside the box, press **`Enter`** to run the rest of the code.

![__](https://biologicslab.co/BIO3523/images/class_01/class_01_5_image12A.png)

**IMPORTANT**

Your bot token is checked and then _immediately_ discarded to protect your bot's security. No copy of your token is kept on the course file server.

````python
# @title Exercise 1: Verify Your Telegram Bot Token

import re
import hashlib
from getpass import getpass
from IPython.display import Image, display

# ── INPUT SECTION ──────────────────────────────────────────────────
bot_token_ex = getpass("Use CTRL-V or CMD-V to paste your Telegram bot token here and hit Enter: ")
# ─────────────────────────────────────────────────────────────────

# Compute and print MD5 hash of the Bot token
md5_hash_ex = hashlib.md5(bot_token_ex.encode("utf-8")).hexdigest()
print("MD5 hash of your Bot token:", md5_hash_ex)

# A Telegram bot token has the form  <bot_id>:<auth_string>
#   - bot_id      : the bot's numeric ID (all digits)
#   - auth_string : ~35 URL-safe characters (letters, digits, '_' and '-')
pattern_ex = re.compile(r'^\d+:[A-Za-z0-9_-]{30,}$')
stripped_token_ex = bot_token_ex.strip()
is_valid_ex = bool(pattern_ex.match(stripped_token_ex))

if is_valid_ex:
    bot_id_ex = stripped_token_ex.split(':')[0]
    is_valid_ex = bot_id_ex.isdigit()

if is_valid_ex:
    print("Your Telegram Bot token is Valid")
    display(Image(url="https://biologicslab.co/BIO3523/images/class_01/congratulations-congrats.gif"))
else:
    print("Your Telegram Bot token is not Valid. \nPlease try again.\n")
    print("Hints:")
    print("Copy-and-paste your Telegram Bot token from your text file into the entry box and hit Enter.")
    print("A valid token looks like  123456789:AAExample...  (digits, a colon, then ~35 characters).")
    print("If you are still having problems, please ask your Instructor for help.")
````
