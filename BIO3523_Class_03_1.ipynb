## **The `/model` Command**

The **`/model`** command in Hermes Agent allows users to switch between already-configured models mid-session without restarting or losing conversation context. It provides a seamless way to change active LLM backends or specific model versions on the fly.

#### **Use Cases**

* **Mid-Session Model Switching:** Instantly shift to a different model (e.g., `/model claude-sonnet-4`) when a particular task demands specialized reasoning strengths or faster turnaround.
* **Persisting Global Defaults:** Update and save your default model preference globally across future sessions using configuration flags like `--global` (e.g., `/model gpt-5 --global`).
* **Task-Specific Optimization:** Dynamically adjust your backend model depending on the complexity of the current prompt without needing to close or re-initialize your terminal session.

Run the `/model` command in your Hermes Agent CLI:
```type
/model Qwen/Qwen3.8-27B-FP8 --global
```
You should see something _similar_ to this image:

```text
·  /model Qwen/Qwen3.8-27B-FP8 --global
·  model → Qwen/Qwen3.8-27B-FP8
```
**IMPORTANT NOTE:** If a Select reasoning level menu is displayed, you need to select **`xhigh`** for the Qwen model, since `high` is not a valid option.

Since you added the `--global` flag to the command, `Qwen/Qwen3.8-27B-FP8` will become the default model for your Hermes Agent.

## **The `/background` Command**

The **`/background`** (or **`/bg`**) command in Hermes Agent enables you to execute a prompt asynchronously in a separate background process or thread. It launches an independent agent session that inherits your current model and configuration settings while leaving your active conversation timeline completely untouched.

####**Use Cases**

* **Executing Parallel Operations:** Kick off a resource-intensive coding or research task in the background so you don't have to pause your active workflow.
* **Multitasking Without Context Loss:** Keep chatting or issuing commands in your main session while Hermes processes a separate, time-consuming request concurrently.
* **Asynchronous Reporting:** Automatically receive a clean completion summary or Rich panel notification in your interface once the background task finishes execution.



## **The `/steer` Command**

The **`/steer`** command in Hermes Agent provides real-time intervention control while an agent or background task is actively executing. It allows you to redirect, refine, or adjust the agent's instructions on the fly without needing to completely abort the current run and start over.

####**Use Cases**
* **Course Correction Mid-Execution:** Instantly pivot an agent's focus if it starts drifting down an inefficient path during a complex, multi-step coding or research task.
* **Injecting Constraints:** Add new parameters or safety rules (e.g., restricting file paths or output formats) while a long-running background process is actively generating results.
* **Saving Compute Resources:** Guide an ongoing model generation back on track immediately rather than waiting for completion, reviewing errors, and re-running the entire prompt.


## **The `/compress` Command**

The **`/compress`** command in Hermes Agent is designed to manage context length by summarizing older conversation history. It reduces token usage across long sessions while retaining critical instructions, code blocks, and context necessary for the agent to continue working effectively.

####**Use Cases**

* **Preventing Context Overflow:** Slim down extended debugging or coding sessions when approaching token window limits without needing to start a completely new session.
* **Streamlining Long-Running Tasks:** Condense verbose back-and-forth chatter into a concise summary state so the agent stays focused on active objectives.
* **Preserving Critical Memory:** Maintain essential project goals and prior decisions while clearing out outdated intermediary thoughts or repetitive outputs.

Run the `/compress` command in the CLI:
```text
/compress --preview
```

## **The `/undo` Command**

The **`/undo`** command in Hermes Agent allows you to instantly revert the most recent interaction, prompt response, or state change within your active session. It acts as a safety net, letting you step backward in the conversation timeline whenever an agent's output heads in an unintended direction.

## Use Cases
* **Reversing Misguided Actions:** Instantly roll back an incorrect file modification or faulty code block generated by the agent during an automated task.
* **Retrying with New Prompts:** Erase an unsatisfactory response and its corresponding prompt so you can rephrase your instructions without cluttering your session history.
* **Error Recovery:** Quickly undo experimental multi-step workflows or command executions before they compound into larger issues within your workspace.

## **The `/clear` Command**

The **`/clear`** command in Hermes Agent completely wipes the current session's conversation history and resets the working memory. It provides a clean slate, allowing you to start fresh without closing out your active terminal window or losing your loaded profile and model configurations.

####**Use Cases**

* **Starting a New Project Focus:** Wipe away old debugging chatter or unrelated prompts when pivoting to an entirely different coding or research task within the same session.
* **Clearing Context Bloat:** Instantly resolve lingering context pollution or conflicting instructions that might be throwing off the agent's reasoning.
* **Quick Session Reset:** Reset your workspace interface in a single keystroke instead of manually exiting and relaunching the Hermes Agent environment.

####**Use Cases**
