In this lessson we continue our exploration of using Hermes Agent to vibe code software applications ("apps"). Vibe coding simply means using normal language to tell Hermes Agent what kind of app you want it to build. Whether you know it or not, this is a **huge deal!**

Vibe coding only became possible within the last year. Before that, software applications required teams of highly trained programmers working many months (or years) to create a commerically viable software program. After "alpha testing" and "beta testing" the software would be released to the general public. This is exactly how the software that you use everyday (MS Windows, macOS, MS Word, Google Colab, GMAIL, etc) was created.

For obvious reasons, no one asked **_you_** exactly what features or functions that **_you_** wanted in a particular app. For better or worse, you just had to accept whatever functions and features the program offered.

NCBI recommends an email address but doesn't enforce it. Requests without one work fine. The email only lets NCBI contact you if your traffic causes a problem. Without it you lose that courtesy, but nothing breaks, and the tool=pubmed_watcher parameter still identifies the app.

NCBI also recommends that you have an NCBI API key. This is also optional, but not having an NCBI API key does have a cost. Without a key, NCBI allows about 3 requests per second per IP, and all your containers likely share campus egress.

A single student's run is fine, but 35 running at once may hit **429 errors**. There are two ways around this issue. You could create your own NCBI account and get a free NCBI API key. ### **OpenRouter Content Filtering**

When utilizing an aggregator and API gateway like OpenRouter, content filtering typically occurs across two layers: OpenRouter’s own client-side or gateway safety rules (such as built-in PII, secret, and regex-based prompt injection detectors) and the native safety guidelines enforced by the underlying model provider or hosting partner processing the request (e.g., OpenAI, Anthropic, Google, or specialized open-weight hosting clusters).

While exact thresholds vary depending on the specific model architecture selected, mainstream LLM providers universally block requests touching specific core safety and security categories:

* **Harmful or Dangerous Activities:** Instructions, blueprints, or step-by-step assistance for creating weapons, cyberattacks, malware, explosives, or illegal acts.

* **Hate Speech & Harassment:** Content that promotes discrimination, disparagement, violence, or targeted bullying against individuals or protected identity groups.

* **Explicit or Sexual Content:** Non-consensual sexual material, erotic roleplay, graphic descriptions of sexual acts, or child sexual abuse material (CSAM).

* **Self-Harm:** Content encouraging, planning, or depicting suicide or physical self-harm.

* **Data Privacy & Credential Leaks:** Prompts attempting to exfiltrate private personal identifiable information (PII), proprietary API keys, secrets, or system passwords.

* **Prompt Injections & Jailbreaks:** System overrides designed to bypass safety filters by tricking the model into ignoring its core behavioral boundaries.

OpenRouter also supports custom content filters and regex rules that developers can configure to automatically intercept or sanitize specific patterns before they reach upstream inference nodes.

------------------

A filovirus paper being refused is a realistic example of biosafety-adjacent text tripping an automated filter, which is a real hazard for anyone building literature tools in infectious disease.

Unfortunately, for biomedical researchers, content filter can occur frequently when working with medical information.

But here is where using `anthropic/claude-sonnet-5` model paid off. Here is the ugly truth. Many models in this situation would simply invent a summary to "replace" the missing one. This is called **hallucination**. Models hallucinate all the time. Does that mean you can't trust what a model tells you?

Well here is the honest answer:

##### **NEVER TRUST WHAT A MODEL TELLS YOU!**

Models hallucinate. They are trained to provide answers. When an answer is missing, they will simply make-up an answer for you.

While Anthropic models are not perfect, this company is known for creating models with comparatively low hallucination rates. Sonnet-5 handling of the situation was right: it caught the null response, reported it plainly, and didn't invent a summary.
